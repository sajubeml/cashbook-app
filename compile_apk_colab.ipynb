### Step 3: Download the Generated APK

```python
from google.colab import files
import glob

apks = glob.glob('bin/*.apk')
if apks:
    print(f'Downloading {apks[0]}...')
    files.download(apks[0])
else:
    print('No APK found in bin/ directory.')
```
